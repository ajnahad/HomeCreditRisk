## Exploratory Data Analysis 

```python
# Dependencies
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
```

```python
folder_path = Path('data/home-credit-default-risk')
figures_folder = Path('figures')

csv_count = sum(1 for _ in folder_path.glob('*.csv'))
print(f'There are {csv_count} files in {folder_path.name}. \n')

for file_path in folder_path.glob('*.csv'):
    heatmap_file = figures_folder / f'{file_path.stem}_missing_values_heatmap.png'

    if heatmap_file.exists():
        print(f'File {heatmap_file.name} has been skipped (already processed.)')
        continue
    try:
        df = pd.read_csv(file_path, encoding='latin_1')
    except UnicodeDecodeError:
        df = pd.read_csv(file_path, encoding='cp1252')
    # First look
    print(f'{file_path.name} has been successfully read into a DataFrame. Lenght: {len(df)}\n')
    print(f'Columns: {df.columns.tolist()}')
    print(f'Basic Stats: {df.describe()}\n {df.info()}')
    # Missing values heatmap
    plt.figure(figsize=(15, 15))
    sns.heatmap(df.isna(), cbar=False, cmap='viridis') # can't use histogram as sns removes missing vals automatically
    plt.title('Missing Values Heatmap')
    plt.savefig(f'figures/{file_path.stem}_missing_values_heatmap.png')
    plt.close()
```
